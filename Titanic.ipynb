first, we will import pandas and read the dataset

```python

import pandas as pd
df = pd.read_csv("/content/Titanic-Dataset.csv")
df.head()
```

```python
df.columns #identifying the columns
```

```python
df.info
```

```python
df.Survived
```

```python
#counting the number of passengers who survived and passengers who died
df['Survived'].value_counts()
```

```python
df.describe()
```

```python
df['Age'].isnull().sum()
#Important for getting the mean, min, max values with the count of each column
```

```python
df['Age']=df['Age'].fillna(df['Age'].median(), inplace=True)
### Handling Missing Values in Age
#The `Age` column contained missing values. These were filled using the median value.
#The median was chosen instead of the mean because the dataset may contain outliers, and the median is more robust to extreme values.
```

```python
df.info()
```

```python
#Now let's see the 'cabin' column
df['Cabin'].isnull().sum()
```

```python
# since many missing values, it will be beneficiary for us to just ignore the column completely
df.drop(columns='Cabin', inplace=True)
```

```python
df.info()
```

```python
df['Embarked']=df['Embarked'].isnull().sum()
```

```python
#since only 2 missing values, we can fill it with the most frequent value(mode)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
```

```python
df.isnull().sum()
```

```python
df['Survived'].value_counts(normalize=True)
```

survived = 0.383838 and not survived = 0.616162
overall survival rate is approximately 0.383838

```python
df.groupby('Sex')['Survived'].mean()
print(742038-188908)
```

The survival rate for female passengers was approximately 74.2%, compared to 18.9% for male passengers, indicating a significantly higher likelihood of survival among females (a difference of ~55 percentage points).

```python
df.groupby('Pclass')['Survived'].mean()
```

- Passengers of upper class had more chance of surviving, indicating the socio-economic advantage.

```python
df.groupby(['Sex','Pclass'])['Survived'].mean()
```

- Females had significantly more rate of survival than man in all classes, suggesting the priority of women in rescues. Additionally, passengers of higher classes showed more rate of survival than the lower classes, suggesting a socio-economic advantage.

```python
df.groupby('Embarked')['Survived'].mean()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")
```

```python
plt.figure(figsize=(8,5))
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Survived', data=df)
plt.title('Survival Count')
plt.show()
```

```python
sns.countplot(x='Sex', data=df)
plt.title('Survival Count by Gender')
plt.show()
```

```python
sns.countplot(x='Pclass', data=df)
plt.title('Survival Count by Passenger Class')
plt.show()
```

```python
sns.histplot(df['Age'],bins=20, kde=True)
plt.title('Age distribution of Passengers')
plt.show()
```

```python
sns.histplot(df['Fare'], bins=10, kde=True)
plt.title('Fare distribution of Passengers')
plt.show()
```

```python
plt.figure(figsize=(8,5))
df['AgeGroup'] = pd.cut(df['Age'], bins=[0, 12, 18, 35, 60, 100], labels=['Child', 'Teen', 'Young Adult', 'Adult', 'Senior'])

sns.countplot(x='AgeGroup',hue='Survived', data=df)
plt.title('Survival Count by Age')
plt.show()
```

```python
sns.barplot(x='Pclass', y='Survived', hue='Sex', data=df)
plt.title('Survival by Class and Gender')
plt.show()
```
